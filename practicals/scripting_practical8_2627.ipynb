#### Warm-up 1 (classes, practical 5)

Write a class `Penguin` with attributes `species`, `island` and `body_mass` (in grams), and a method `is_heavy()` that returns `True` if the penguin weighs more than 4000 grams. Create two penguins, one heavy and one light, and test the method on both. Later in this session, the same kind of data comes as a table with 344 penguins.

```python
# insert your code here
class Penguin:
  def __init__(self,species,island,body_mass):
    self.species = species
    self.island = island
    self.body_mass = body_mass

  def is_heavy(self):
    if self.body_mass > 4000:
      return True

p1 = Penguin('Adelie', 'Dream', 3500)
p2 = Penguin('Gentoo', 'Biscoe', 5000)

print(p1.is_heavy())
print(p2.is_heavy())
```

#### Warm-up 2 (counting with dictionaries, practical 3)

The list below contains island names. Count how often each island occurs, using the counting-with-a-dictionary idiom. Later in this session you will see that pandas can do this in a single line.

```python
islands = ['Biscoe', 'Dream', 'Biscoe', 'Torgersen', 'Dream', 'Biscoe']
# insert your code here

count_island = {}

for i in islands:
  if i in count_island:
    count_island[i]+=1
  else:
    count_island[i]=1

print(count_island)
```

## Selecting slices of rows/columns

There are two ways to extract slices of rows/columns from your `DataFrame`. You can either use `loc`, which is used to select rows/columns by label name; or you can use `iloc`, which selects by row/column index:

```python
df.loc[:,'Species':'Island']
```

And based on the index:

```python
df.iloc[:,2:5]
```

## Subframes based on specific row values

By creating a Boolean series object, it is possible to filter a `DataFrame` based on specific values.

### Creating Boolean Series

A Boolean series object, which checks whether a certain condition is true for a particular instance, can be created as follows:

```python
df['Island'] == 'Torgersen'
```

### Boolean series as filter

Such a Boolean filter can then be used in order to filter instances from the dataframe based on that condition:

```python
df.loc[df['Island'] == 'Torgersen' ]
```

We can equally combine conditions in order to perform a more complex filtering based on multiple features:

```python
df.loc[(df['Island'] == 'Torgersen') & (df['Body Mass (g)'] > 4500)]
```

## Sorting

Sorting all instances based on a single feature, from small to large:

```python
df.sort_values('Flipper Length (mm)')
```

And from large to small:

```python
df.sort_values('Flipper Length (mm)', ascending=False)
```

### Challenge 1

Write a function `heaviest(df, island, n)` that returns the `n` heaviest penguins living on a given island, as a DataFrame (so a filter, a sort and `head()`, in one function). Test it with `heaviest(df, 'Dream', 3)`, and check the result against the table.

```python
# insert your code here
def heaviest(df,island,n):
  return df.loc[df['Island'] == island,:]\
      .sort_values('Body Mass (g)', ascending = False)\
      .head(n)

print(heaviest(df, 'Dream', 3))
```

### Challenge 2

The method `unique()` of a Series gives the different values that occur in it: `df['Species'].unique()`. Use it in a `for` loop over the species, and print for every species its name and the mean body mass of its penguins. Next week we will see that pandas can do this in one line, with `groupby`.

```python
# insert your code here
species = df['Species'].unique()

for specie in species:
  weight_mean = df.loc[df['Species'] == specie, 'Body Mass (g)'].mean()
  print(specie,weight_mean)
```
